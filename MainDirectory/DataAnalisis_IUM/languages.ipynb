# Languages data analysis

A quick breakdown of the languages dataset.

```python
import pandas as pd
from matplotlib import pyplot as plt
import plotly.express as px
df_languages = pd.read_csv('data/languages.csv')
df_languages.head()
```

- Display all the languages used in the dataset

```python
languages = df_languages['language'].unique()
print("Number of languages: ", len(languages))
for language in languages:
    print(language)
```

- Display the number of movies for each language, (top 10)

```python
df_grouped = df_languages['language'].value_counts().reset_index()
df_grouped.columns = ['Language', 'Number of Movies']
df_grouped.head(10)
```

### Visualizations by chart
- Pie chart
- Bar chart

```python

fig, ax = plt.subplots(1,2 , figsize=(30, 40))

df_grouped.set_index('Language').head(10).plot(kind='pie', y='Number of Movies', autopct='%1.1f%%', legend= False, figsize=(15, 15), ax=ax[0])
df_grouped.head(10).plot(kind='bar', x='Language', y='Number of Movies', legend=True, color='purple', figsize=(13, 9), ax=ax[1])
ax[1].set_ylabel('Number of Movies')
ax[1].set_xlabel('Language')
plt.ylabel('')  # Rimuove l'etichetta dell'asse Y
plt.title('Distribution of movies by language')
plt.tight_layout()
plt.show()
```

### Interactive charts
now we will create interactive charts using Plotly, displying the top 12 languages.

```python

fig = px.bar(df_grouped.head(12), x='Language', y='Number of Movies', title='Numero di Film per Lingua')
fig.show()
```
