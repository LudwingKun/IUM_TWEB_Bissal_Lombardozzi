Are there any missing or duplicated values?

```python
duplicates = df_movies['id'].duplicated().sum()
print("Duplicated films: " + str(duplicates))

missing = df_movies['rating'].isnull().sum()
print("Missing ratings: " + str(missing))
```

Does the length of a movie affect its rating?
Yes, there is a weak positive correlation between the length of a movie and its rating.
The longer the movie, the higher the rating.

```python
sns.regplot(data=df_movies, x='minute', y='rating')
plt.ylim(0, 6)
plt.xlim(0, 3000)
```
