
# Movies data analysis

```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns

df_movies = pd.read_csv('../data/movies.csv')
df_movies.head()
```

Are there any missing or duplicated values?

```python
duplicates = df_movies['id'].duplicated().sum()
print("Duplicated films: " + str(duplicates))

missing = df_movies['rating'].isnull().sum()
print("Missing ratings: " + str(missing))
```

Does the length of a movie affect its rating?
Yes, there is a weak positive correlation between the length of a movie and its rating.

```python
sns.regplot(data=df_movies, x='minute', y='rating')
plt.ylim(0, 6)
plt.xlim(0, 3000)
```

```python
df_films_per_year = df_movies.groupby('date').size().reset_index(name='count')

sns.lineplot(data=df_films_per_year, x='date', y='count')
plt.xticks(range(1900, 2023, 5), rotation=90)
plt.title('Number of Movies per Year')
plt.xlabel('Year')
plt.ylabel('Number of Movies')
plt.show()
```
